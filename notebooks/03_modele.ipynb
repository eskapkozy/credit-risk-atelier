```python
import sys
sys.path.insert(0, "..")
import pandas as pd
from src.features import preparer
df = pd.read_csv("../data/processed/credit_propre.csv")
X, y = preparer(df)
X.shape, round(y.mean(), 3)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
logit = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(X_train, y_train)
print("AUC régression logistique :", round(roc_auc_score(y_test, logit.predict_proba(X_test)[:, 1]), 3))
```
