```python
import pandas as pd
df = pd.read_csv("../data/raw/credit_risk_dataset.csv")
print(df.isna().sum())
print("doublons :", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
df = df[(df.person_age <= 100) & (df.person_emp_length.isna() | (df.person_emp_length <= 60))]
df["person_emp_length"] = df["person_emp_length"].fillna(df["person_emp_length"].median())
df["loan_int_rate"] = df.groupby("loan_grade")["loan_int_rate"].transform(lambda s: s.fillna(s.median()))
print(df.shape, "valeurs manquantes :", df.isna().sum().sum())
```

```python
import os
os.makedirs("../data/processed", exist_ok=True)
df.to_csv("../data/processed/credit_propre.csv", index=False)
os.listdir("../data/processed")
```
